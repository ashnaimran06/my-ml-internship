## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

### Split design: Grouped train/test split

I will use a grouped train/test split based on `client_hash_id`. Around 80% of clients will be used for training and 20% for testing.

This is a suitable split because multiple content pages can belong to the same client. If pages from one client appear in both training and testing, the model may learn client-specific patterns and produce overly optimistic results.

Keeping each client's pages in only one split gives a more honest test of whether the model can rank pages for clients it has not seen before.

I will use a fixed random seed of 42 to make the split reproducible. The W04 baseline and the Decision Tree will be evaluated on the exact same test pages using Precision@20 and Precision@50.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


from sklearn.model_selection import GroupShuffleSplit

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

print("Split method: GroupShuffleSplit")
print("Test size: 20% of clients")
print("Random state: 42")
```
